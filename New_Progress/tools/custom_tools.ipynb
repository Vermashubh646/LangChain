### Using Pydantic

```python
from pydantic import BaseModel, Field
from typing import Annotated
from langchain_core.tools import StructuredTool
```

```python
class Number_mul(BaseModel):
    a:Annotated[int, Field(...,description="The first number to add")]
    b:Annotated[int, Field(...,description="The second number to add")]
```

```python
def multiply_func(a: int,b :int)->int:
    return a*b
```

```python
multiply_func = StructuredTool.from_function(
    func = multiply_func,
    name="multiply",
    description="Multiply two numbers",
    args_schema=Number_mul
)
```

```python
result=multiply_func.invoke({'a':15,'b':656542})

print(result,'\n\n')
print(multiply_func.name,'\n\n')
print(multiply_func.description,'\n\n')
print(multiply_func.args,'\n\n')
```

### For Production related

```python
from langchain.tools import BaseTool
from typing import Type
from typing import Annotated
```

```python
# arg schema using pydantic

class MultiplyInput(BaseModel):
    a:Annotated[int, Field(...,description="The first number to add")]
    b:Annotated[int, Field(...,description="The second number to add")]
```

```python
class MultiplyTool(BaseTool):
    name: str = "multiply"
    description: str = "Multiply two numbers"

    args_schema: Type[BaseModel] = MultiplyInput

    def _run(self, a: int, b: int) -> int:
        return a * b
```

```python
multiply_tool = MultiplyTool()
```

```python
result = multiply_tool.invoke({'a':3, 'b':3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)

print(multiply_tool.args)
```

## Toolkit

```python
from langchain_core.tools import tool
              
# Custom tools
@tool
def add(a: int, b: int) -> int:
    """Add two numbers"""
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b
```

```python
class MathToolkit:
    def get_tools(self):
        return [add, multiply]
```

```python
toolkit = MathToolkit()
tools = toolkit.get_tools()

for tool in tools:
    print(tool.name, "=>", tool.description)
```
